```python
FILE = "data/github-repos-url.txt"

repos_urls = []
with open(FILE, 'r') as f:
    repos_urls = [line.strip() for line in f]
repos_urls
```

```python
import requests


def get_issue_data(repo_url):
    url = repo_url + "/issues"
    headers = {
        "Accept": "application/vnd.github.v3+json",
        "Content-Type": "application/json",
        'User-Agent': 'XY',
    }
    # Añade la autorización si se proporciona un token.
    # if GITHUB_TOKEN:
    #     headers["Authorization"] = f"token {GITHUB_TOKEN}"
    params = {
        'state': 'all',  # Cambia a 'all' para incluir cerradas
        'per_page': 100,
        'page': 1
    }

    all_issues = []
    print(f"Buscando issues para [{url}]")

    while True:
        response = requests.get(url, headers=headers, params=params)

        if response.status_code == 200:
            issues = response.json()

            if not issues:
                # No hay más issues, salimos del bucle.
                break

            for issue in issues:
                # Los Pull Requests también se listan como issues,
                # pero tienen una clave 'pull_request'.
                # Si queremos excluirlos:
                if 'pull_request' in issue:
                    continue

                # Extraemos el título y la descripción (body)
                all_issues.append({
                    "number": issue.get("number"),
                    "title": issue.get("title"),
                    "description": issue.get("body")
                })

            print(f"-> Issues obtenidas en la página {params['page']}: {len(issues)}")
            params['page'] += 1  # Prepara para la siguiente página

        elif response.status_code == 404:
            print(f"Error: Repositorio '{url}' no encontrado (código 404). Verifica los nombres.")
            return None

        elif response.status_code == 403:
            # Límite de tasa (rate limit) alcanzado o token inválido
            print(
                "Error: Límite de tasa de GitHub alcanzado o Token de Acceso Personal inválido/insuficiente (código 403).")
            # Para más detalles:
            print(response.headers.get("X-RateLimit-Remaining"))
            return None

        else:
            print(f"Error inesperado al acceder a la API: Código {response.status_code}")
            print(response.text)
            return None

    return all_issues
```

```python
all_respos_issues = []

for repo_url in repos_urls:
    all_respos_issues.append(get_issue_data(repo_url))

all_respos_issues
```
